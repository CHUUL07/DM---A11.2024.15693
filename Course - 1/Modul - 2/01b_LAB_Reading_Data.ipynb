```python
# Imports
import sqlite3
import urllib.request
from pathlib import Path

import pandas as pd
```

## Latihan: membaca berkas basis data

Kerjakan langkah berikut pada basis data `baseball.db`:

1. Buat variabel `path` yang menunjuk ke berkas `baseball.db` di dalam folder `data/`.
2. Buat koneksi `con` ke basis data pada `path` tersebut.
3. Buat variabel `query` berisi perintah SQL untuk membaca seluruh isi tabel `allstarfull`.
4. Muat hasilnya menjadi DataFrame `observations` memakai `pandas.read_sql`.

### Tambahan (opsional)

- Simpan daftar tabel pada `sqlite_master` ke variabel `tables`.
- Bayangkan kamu sedang menyusun daftar pemain terbaik sepanjang masa. Urutkan pemain berdasarkan jumlah pertanyaan all-star, lalu ambil tiga nama teratas.


```python
# Unduh basis data bila belum tersedia
database_path = Path("data/baseball.db")
database_url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-ML0232EN-SkillsNetwork/asset/baseball.db"

if not database_path.exists():
    database_path.parent.mkdir(parents=True, exist_ok=True)
    urllib.request.urlretrieve(database_url, database_path)
```

```python
# Buka koneksi ke basis data
path = database_path
con = sqlite3.connect(path)

def read_query(sql, connection=con):
    """Jalankan query SQL dan kembalikan hasilnya sebagai DataFrame."""
    return pd.read_sql(sql, connection)

query = """
SELECT *
FROM allstarfull;
"""

observations = read_query(query)
observations.head()
```

```python
# Daftar tabel yang tersedia pada basis data
tables = read_query("SELECT name, type FROM sqlite_master")
tables
```

```python
# Tiga pemain dengan penampilan all-star terbanyak
best_query = """
SELECT playerID,
       SUM(GP) AS num_games_played,
       AVG(startingPos) AS avg_starting_position
FROM allstarfull
GROUP BY playerID
ORDER BY num_games_played DESC, avg_starting_position ASC
LIMIT 3;
"""

best_players = read_query(best_query)
best_players
```

## Pembahasan

Tabel `allstarfull` mencatat setiap penampilan pemain pada pertandingan all-star. Kolom `GP` berisi jumlah pertandingan yang dimainkan, sedangkan `startingPos` menunjukkan posisi awal pemain. Dengan mengelompokkan data per `playerID`, menjumlahkan `GP`, lalu mengurutkannya dari yang terbesar, kita memperoleh tiga pemain dengan kontribusi all-star paling banyak. Nilai `avg_starting_position` dipakai sebagai penentu kedua saat jumlah pertandingan sama.
